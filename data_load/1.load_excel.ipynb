## 2.Codfifique variables categoricas usando LabelEnconder y OneHot Encoding


```python
import pandas as pd

# 1. Cargamos el DataFrame (necesario para que 'df' vuelva a existir en esta sesión)
df = pd.read_excel('data/dataset_ventas.xlsx')

# 2. Aplicamos One-Hot Encoding a la columna 'categoria' usando Pandas
df_onehot_pd = pd.get_dummies(df, columns=['categoria'], prefix='cat')

# 3. Mostramos el resultado
print("\n--- Resultado de One-Hot Encoding con Pandas ---")
print(df_onehot_pd[['cat_Alimentos', 'cat_Hogar', 'cat_Papelería', 'cat_Tecnología']].head())
```

## 3.Aplique normalizacion y estandarizacion.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Cargamos el DataFrame por si acaso y seleccionamos las columnas numéricas a escalar
df = pd.read_excel('data/dataset_ventas.xlsx')
columnas_numericas = ['cantidad', 'precio_unitario', 'total_venta']

# --- TÉCNICA 1: NORMALIZACIÓN (MinMaxScaler) ---
# Escala los datos en el rango [0, 1]
scaler_minmax = MinMaxScaler()
df_normalizado = df.copy()
df_normalizado[columnas_numericas] = scaler_minmax.fit_transform(df[columnas_numericas])

print("--- Resultado de Normalización (MinMax: 0 a 1) ---")
print(df_normalizado[columnas_numericas].head())

# --- TÉCNICA 2: ESTANDARIZACIÓN (StandardScaler) ---
# Centra los datos con media 0 y desviación estándar 1
scaler_standard = StandardScaler()
df_estandarizado = df.copy()
df_estandarizado[columnas_numericas] = scaler_standard.fit_transform(df[columnas_numericas])

print("\n--- Resultado de Estandarización (Media=0, Desviación=1) ---")
print(df_estandarizado[columnas_numericas].head())
```

## 4.Balance la variable objetivo usando SMOTE

```python
%pip install imbalanced-learn
```

```python
import pandas as pd
from imblearn.over_sampling import SMOTE

# 1. Cargamos el DataFrame
df = pd.read_excel('data/dataset_ventas.xlsx')

# 2. Creamos una variable objetivo (target) de ejemplo desbalanceada
# Ejemplo: 1 si es de la categoría 'Tecnología' (clase minoritaria), 0 para las demás
df['target'] = (df['categoria'] == 'Tecnología').astype(int)

print("--- Distribución ANTES de SMOTE ---")
print(df['target'].value_counts())

# 3. Seleccionamos las características (X) y la variable objetivo (y)
# Usamos las variables numéricas principales para el ejemplo
X = df[['cantidad', 'precio_unitario', 'total_venta']]
y = df['target']

# 4. Aplicamos SMOTE para balancear las clases
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

print("\n--- Distribución DESPUÉS de SMOTE ---")
print(pd.Series(y_resampled).value_counts())
```

## 5.Verificar si hay valores dupplicados 


```python
df.duplicated().sum()

total_duplicados = df.duplicated().sum()
print(f"Cantidad de filas duplicadas encontradas: {total_duplicados}")

# 3. (Opcional) Si quieres ver cuáles son las filas duplicadas
if total_duplicados > 0:
    print("\n--- Filas duplicadas detectadas ---")
    print(df[df.duplicated(keep=False)])

# 4. Eliminar filas duplicadas (si las hubiera) y conservar la primera aparición
df_limpio = df.drop_duplicates()

print(f"\nFilas originales: {len(df)}")
print(f"Filas después de eliminar duplicados: {len(df_limpio)}")
```

## 6.Eliminar valores duplicados

```python
# 1. Cargamos el DataFrame por si acaso
df = pd.read_excel('data/dataset_ventas.xlsx')

# 2. Contamos cuántos duplicados había antes de eliminar
duplicados_antes = df.duplicated().sum()
print(f"Filas duplicadas antes de limpiar: {duplicados_antes}")

# 3. Eliminamos las filas duplicadas (mantiene la primera aparición por defecto)
df = df.drop_duplicates()

# 4. Verificamos cuántos duplicados quedan
duplicados_despues = df.duplicated().sum()
print(f"Filas duplicadas después de limpiar: {duplicados_despues}")

# 5. Mostramos el tamaño final del DataFrame limpio
print(f"Dimensiones finales del DataFrame (filas, columnas): {df.shape}")
```
