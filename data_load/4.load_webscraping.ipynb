```python
import requests
import pandas as pd

# 1. Definimos la URL de la API de Rick y Morty ( endpoint de personajes )
url = "https://rickandmortyapi.com/api/character"

# 2. Hacemos la petición GET a la API
response = requests.get(url)

# 3. Verificamos que la conexión sea exitosa (código 200)
if response.status_code == 200:
    # Convertimos la respuesta JSON en un diccionario de Python
    data = response.json()
    
    # Los personajes están dentro de la clave 'results'
    personajes = data['results']
    
    # 4. Creamos el DataFrame de Pandas con la lista de personajes
    df_rick = pd.DataFrame(personajes)
    
    print("--- ¡Datos de la API de Rick y Morty cargados con éxito! ---")
    # Mostramos las primeras filas y columnas principales
    display(df_rick[['id', 'name', 'status', 'species', 'gender']].head())
    
else:
    print(f"Error al conectar con la API: {response.status_code}")
```
