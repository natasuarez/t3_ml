```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
from imblearn.over_sampling import SMOTE

# ==========================================
# 1. Cargar el archivo CSV
# ==========================================
# Asegúrate de cambiar 'data/dataset_ventas.csv' por la ruta o nombre real de tu archivo CSV
df = pd.read_csv('data/dataset_ventas.csv')

print("--- Primeras 5 filas del CSV ---")
display(df.head())


# ==========================================
# 2. Codificar variables categóricas (LabelEncoder y One-Hot Encoding)
# ==========================================
df_encoded = df.copy()

# A) LabelEncoder para la variable 'ciudad'
le = LabelEncoder()
df_encoded['ciudad_encoded'] = le.fit_transform(df_encoded['ciudad'])

# B) One-Hot Encoding para la variable 'categoria' usando Pandas
df_encoded = pd.get_dummies(df_encoded, columns=['categoria'], prefix='cat')

print("\n--- 2. Variables categóricas codificadas ---")
print(df_encoded[['ciudad', 'ciudad_encoded']].head())


# ==========================================
# 3. Normalización y Estandarización
# ==========================================
columnas_numericas = ['cantidad', 'precio_unitario', 'total_venta']

# A) Normalización (MinMaxScaler: escala los valores en el rango [0, 1])
scaler_minmax = MinMaxScaler()
df_normalizado = df.copy()
df_normalizado[columnas_numericas] = scaler_minmax.fit_transform(df[columnas_numericas])

# B) Estandarización (StandardScaler: media 0, desviación estándar 1)
scaler_standard = StandardScaler()
df_estandarizado = df.copy()
df_estandarizado[columnas_numericas] = scaler_standard.fit_transform(df[columnas_numericas])

print("\n--- 3. Datos Normalizados (MinMax) ---")
print(df_normalizado[columnas_numericas].head())


# ==========================================
# 4. Balancear la variable objetivo usando SMOTE
# ==========================================
# Creamos una variable objetivo (target) de ejemplo basada en una categoría
df['target'] = (df['categoria'] == 'Tecnología').astype(int)

X = df[['cantidad', 'precio_unitario', 'total_venta']]
y = df['target']

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

print("\n--- 4. Distribución de la clase objetivo después de SMOTE ---")
print(pd.Series(y_resampled).value_counts())


# ==========================================
# 5. Verificar si hay valores duplicados
# ==========================================
total_duplicados = df.duplicated().sum()
print(f"\n--- 5. Verificación de duplicados ---")
print(f"Cantidad de filas duplicadas encontradas: {total_duplicados}")


# ==========================================
# 6. Eliminar valores duplicados
# ==========================================
df_limpio = df.drop_duplicates()
print(f"\n--- 6. Eliminación de duplicados ---")
print(f"Filas antes de limpiar: {len(df)}")
print(f"Filas después de eliminar duplicados: {len(df_limpio)}")
```
